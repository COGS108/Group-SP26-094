## Research Question

Is a rise in air pollutants accompanied by an increase in the severity (length/cost) of a power outage in the continental United States of America, and do these predictors change by region?

## Background and Prior Work

&emsp;In the United States, air quality is monitored and defined by the Environmental Protection Agency (EPA) nested in the government’s executive branch. Air quality is characterized as the concentration of specific pollutants in the atmosphere and their resulting effects on the health of the public/environment. The EPA’s Air Quality Index (AQI) ranges from “Good” to “Hazardous” where the severity of pollutant diffusion across the atmosphere poses little risk to the public (“Good”) to everyone being increasingly likely to experience adverse effects from its exposure (“Hazardous”). The AQI’s index values intuitively rate air quality by having a higher AQI value meaning the greater the level of air pollution and vice versa.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) It is important to note however that the AQI is a composite measure and takes multiple air pollutants into account rather than just an individual pollutant concentration. 

&emsp; As the relationship between power outages and air pollution is concerned, previous work has been done in Ecuador, specifically in the capital, Quito, and their 2023 and 2024 power outages which saw large scale blackouts for several hours for consecutive months. Researchers Vallejo and their colleagues posit that Quito’s blackouts, “...revealed a significant deterioration in atmospheric pollution, primarily driven by the widespread use of diesel generators, increased vehicular traffic, and adverse meteorological conditions…”<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) Vallejo and colleagues’ work in the context of Ecuador’s 2023 and 2024 power outages reveals an interesting interaction between infrastructural issues such as blackouts and the increased usage of alternative power sources which produce pollutants and deteriorate the atmosphere. Another work that explores the relationship between air pollution and power outages takes a similar look at how Delhi, India and its citizens grapple with power outages through reliance on backup solutions like diesel generators. Yatang Lin and Dana Kassem note that diesel generators “...emit 3.5 times more nitrogen oxides and two times more particulate matter than coal-fired plants for the same amount of energy,” which is worsened by the fact that Delhi’s power outages are locally combatted through “The frequent on-off cycling of backup generators…” which “...further increases fuel consumption and pollution,”<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Lin and Kassem’s findings further display the downstream pollutive effects of power outages and how they influence worsening air quality. This explains a possible reasoning as to why air pollutants and strength of power outage could be correlated. This correlation found in this work is the type of relationships we wish find with our research question. To be subject to frequent blackouts and power outages coupled with a reliance on alternative power sources which negatively impact air quality establishes this relationship wherein power outages can possibly predict worsening air quality/pollution. 

&emsp; An important consideration when discussing air quality involves the health complications that come with exposure to air pollutants. As detailed by California’s Air Resources Board, the reliance on backup generators (which are primarily diesel generators) and their emission of diesel particulate matter (DPM) is considered a toxic air contaminant implicated in the development of cancer and increasing the lung’s susceptibility to damage.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) The ability for air pollution and power outages to influence one another has an impact on health either way through exposure to harmful airborne substances or as how Zhang and colleagues put it, power outages threaten vulnerable populations as they suggest an increased risk of COPD exacerbation, as a result of their dependency on electricity.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) The importance of air quality and power reliability influences on the health of the public cannot be understated.

&emsp; This project examines the relationship between air pollution and power outages within the context of the United States and intends to explore the inverse of the previously established case studies done in Quito and Delhi. We intend to examine if air pollutants are a predictor for the severity (length/cost) of power outages nationally, and if these predictors (specific air pollutants) are subject to variability per US region.

CACES U.S. Air Quality Data - https://www.caces.us/data, requires completing a brief access form. 
This dataset provides annual, county-level, population-weighted ambient concentration estimates for ten pollutants: CO, NO₂, O₃, BC, PM10, PM2.5, SO₂, COA, HOA, and PNC, covering 1979–2020. Key columns are `fips` (county identifier), `pollutant`, `year`, `pred_wght` (the population-weighted concentration estimate), `state_abbr`, `lat`, and `lon`. Estimates are derived from EPA monitoring networks combined with land-use regression models, providing coverage even in areas without monitoring stations. We could aggregate to annual state means using `state_abbr` and `year` as keys to match the geographic resolution of the outage data. 

Purdue Major Power Outage Risks in the U.S. - https://engineering.purdue.edu/LASCI/research-data/outages, direct download, no application required.

```python
## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
import pandas as pd
import os

caces_raw = pd.read_excel('data/00-raw/caces_us_data.xlsx')

print("RAW SHAPE")
print(caces_raw.shape)
print("\nCOLUMNS")
print(caces_raw.columns.tolist())
print("\nFIRST 5 ROWS")
caces_raw.head()
```

```python
# Tidy check
# Each row = one county × one year × one pollutant → already tidy (long format)
print("One row per (fips, year, pollutant)?")
dupes = caces_raw.duplicated(subset=['fips','year','pollutant']).sum()
print(f"Duplicate (fips, year, pollutant) rows: {dupes}")

print(f"\nPollutants present: {sorted(caces_raw['pollutant'].unique())}")
print(f"Year range: {caces_raw['year'].min()} – {caces_raw['year'].max()}")
print(f"Unique states: {caces_raw['state_abbr'].nunique()}")
print(f"Unique counties (FIPS): {caces_raw['fips'].nunique()}")
```

```python
# Missing data 
print("MISSING VALUES")
miss = caces_raw.isnull().sum()
miss_pct = (miss / len(caces_raw) * 100).round(2)
missing_df = pd.DataFrame({'missing_count': miss, 'missing_pct': miss_pct})
print(missing_df[missing_df['missing_count'] > 0])

# Check if missingness is systematic by pollutant
print("\nMISSING pred_wght BY POLLUTANT")
print(
    caces_raw[caces_raw['pred_wght'].isna()]
    .groupby('pollutant')
    .size()
    .rename('missing_count')
)
```

```python
# Outlier check on pred_wght 
print("pred_wght SUMMARY BY POLLUTANT")
print(
    caces_raw.groupby('pollutant')['pred_wght']
    .describe()
    .round(4)
    .to_string()
)

# Flag extreme outliers (> 5 IQR from median) per pollutant
def flag_outliers(group):
    q1, q3 = group['pred_wght'].quantile([0.25, 0.75])
    iqr = q3 - q1
    return group[
        (group['pred_wght'] < q1 - 5 * iqr) |
        (group['pred_wght'] > q3 + 5 * iqr)
    ]

outliers = caces_raw.groupby('pollutant', group_keys=False).apply(flag_outliers)
print(f"\nExtreme outliers (>5 IQR): {len(outliers)} rows")
print(outliers[['fips','state_abbr','year','pollutant','pred_wght']]
      .sort_values('pred_wght', ascending=False).head(10))
```

```python
# Clean & aggregate to state × year 
caces = caces_raw.copy()
caces.columns = caces.columns.str.strip().str.lower()

# Restrict to the 6 well-covered criteria pollutants
# (BC, COA, HOA, PNC excluded after coverage check — see merge section)
POLLUTANTS = ["co", "no2", "o3", "pm10", "pm25", "so2"]
caces_filtered = caces[caces['pollutant'].str.lower().isin(POLLUTANTS)].copy()
caces_filtered['pollutant'] = caces_filtered['pollutant'].str.lower()

# Aggregate to state × year: population-weighted mean across counties
state_yr_long = (
    caces_filtered
    .groupby(['year', 'state_abbr', 'pollutant'])['pred_wght']
    .mean()
    .reset_index()
)

# Pivot wide: one column per pollutant
state_yr = (
    state_yr_long
    .pivot_table(index=['year','state_abbr'], columns='pollutant', values='pred_wght')
    .reset_index()
)
state_yr.columns.name = None
state_yr = state_yr.rename(columns={p: f'poll_{p}' for p in POLLUTANTS})

# Normalise keys
state_yr['state_abbr'] = state_yr['state_abbr'].str.strip().str.upper()
state_yr['year'] = state_yr['year'].astype(int)

print("PROCESSED CACES SHAPE")
print(state_yr.shape)
print(f"\nYears: {state_yr['year'].min()} – {state_yr['year'].max()}")
print(f"States: {state_yr['state_abbr'].nunique()}")
print("\nFIRST 5 ROWS")
state_yr.head()
```

```python
# ── Write to processed ────────────────────────────────────────
os.makedirs('data/02-processed', exist_ok=True)
state_yr.to_excel('data/02-processed/caces_state_year.xlsx', index=False)
print("Saved: data/02-processed/caces_state_year.xlsx")
```

```python
## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
# The Purdue Excel file has metadata rows above the header; skiprows=5
# removes them. Adjust if your copy differs — check raw output below.
raw = pd.read_excel(
    'data/00-raw/power_outage_data.xlsx',
    sheet_name=0,
    skiprows=5,
    header=0
)
raw.columns = raw.columns.str.strip()

print(" RAW SHAPE ")
print(raw.shape)
print("\nFIRST 3 ROWS, FIRST 8 COLUMNS")
print(raw.iloc[:3, :8].to_string())
print("\nALL COLUMNS")
print(raw.columns.tolist())
```

```python
# Tidy check & header fix
# Remove the units row that Purdue includes directly below the header
outage = raw.copy()
outage = outage[pd.to_numeric(outage['YEAR'], errors='coerce').notna()].copy()

print(f"Rows after removing non-data rows: {len(outage)}")

# Each row = one outage event → already tidy
dupes = outage.duplicated().sum()
print(f"Fully duplicate rows: {dupes}")
```

```python
# Rename columns
col_map = {
    'YEAR'                    : 'year',
    'MONTH'                   : 'month',
    'U.S._STATE'              : 'state',
    'POSTAL.CODE'             : 'state_abbr',
    'NERC.REGION'             : 'nerc_region',
    'CLIMATE.REGION'          : 'climate_region',
    'CLIMATE.CATEGORY'        : 'climate_category',
    'ANOMALY.LEVEL'           : 'anomaly_level',
    'CAUSE.CATEGORY'          : 'cause_category',
    'CAUSE.CATEGORY.DETAIL'   : 'cause_detail',
    'OUTAGE.DURATION'         : 'outage_duration_min',
    'DEMAND.LOSS.MW'          : 'demand_loss_mw',
    'CUSTOMERS.AFFECTED'      : 'customers_affected',
    'RES.PRICE'               : 'price_residential',
    'COM.PRICE'               : 'price_commercial',
    'IND.PRICE'               : 'price_industrial',
    'TOTAL.PRICE'             : 'price_total',
    'TOTAL.SALES'             : 'total_sales_mwh',
    'TOTAL.CUSTOMERS'         : 'total_customers',
    'PC.REALGSP.STATE'        : 'real_gsp_usd',
    'PC.REALGSP.CHANGE'       : 'real_gsp_change',
    'POPULATION'              : 'population',
    'POPPCT_URBAN'            : 'pct_urban_pop',
    'POPDEN_URBAN'            : 'pop_density_urban',
    'OUTAGE.START.DATE'       : 'outage_start_date',
    'OUTAGE.RESTORATION.DATE' : 'outage_restore_date',
}
col_map = {k: v for k, v in col_map.items() if k in outage.columns}
outage = outage.rename(columns=col_map)

# Numeric coercion 
num_cols = [
    'year', 'month', 'outage_duration_min', 'customers_affected',
    'demand_loss_mw', 'price_residential', 'price_commercial',
    'price_industrial', 'price_total', 'total_sales_mwh',
    'total_customers', 'real_gsp_usd', 'population', 'pop_density_urban',
]
for c in num_cols:
    if c in outage.columns:
        outage[c] = pd.to_numeric(outage[c], errors='coerce')

outage = outage.dropna(subset=['state', 'year'])
outage['year'] = outage['year'].astype(int)
outage['state_abbr'] = outage['state_abbr'].str.strip().str.upper()
outage['outage_duration_hr'] = outage['outage_duration_min'] / 60

print("")
print(outage.shape)
print(f"\nYears: {outage['year'].min()} – {outage['year'].max()}")
print(f"States: {outage['state_abbr'].nunique()}")
outage[['year','state','state_abbr','cause_category',
        'outage_duration_min','price_total']].head()
```

```python
# Missing data
print("MISSING VALUES (columns with any missing)")
miss = outage.isnull().sum()
miss_pct = (miss / len(outage) * 100).round(2)
missing_df = pd.DataFrame({'missing_count': miss, 'missing_pct': miss_pct})
print(missing_df[missing_df['missing_count'] > 0].sort_values(
    'missing_pct', ascending=False))

# Is duration missingness related to cause category?
print("\nDURATION MISSINGNESS BY CAUSE CATEGORY")
print(
    outage.groupby('cause_category')['outage_duration_min']
    .apply(lambda x: x.isna().sum())
    .rename('missing_duration_count')
    .sort_values(ascending=False)
)
# Justification: we do NOT drop rows missing duration here.
# These rows still contribute cause category, customer, and
# regional information. We only drop duration-missing rows
# when duration is specifically the outcome variable in modeling.
```

```python
# Outlier check
print("OUTAGE DURATION SUMMARY")
print(outage['outage_duration_min'].describe().round(1))

# Flag extreme durations (> 5 IQR)
q1 = outage['outage_duration_min'].quantile(0.25)
q3 = outage['outage_duration_min'].quantile(0.75)
iqr = q3 - q1
extreme = outage[outage['outage_duration_min'] > q3 + 5 * iqr]
print(f"\nExtreme duration outliers (>5 IQR): {len(extreme)} events")
print(extreme[['year','state','cause_category',
               'outage_duration_min','customers_affected']]
      .sort_values('outage_duration_min', ascending=False).head(10))

# Note: these are real catastrophic events (hurricanes, ice storms),
# not data errors. We retain them but will log-transform duration
# in modeling to reduce their leverage.
print("\nCUSTOMERS AFFECTED SUMMARY")
print(outage['customers_affected'].describe().round(0))
```

```python
# Write interim (event-level, cleaned)
os.makedirs('data/01-interim', exist_ok=True)
outage.to_excel('data/01-interim/outage_cleaned.xlsx', index=False)
print("Saved: data/01-interim/outage_cleaned.xlsx")

# Aggregate to state × year for merging 
outage_agg = (
    outage
    .groupby(['year', 'state_abbr'])
    .agg(
        n_outages           = ('state_abbr',          'count'),
        total_duration_min  = ('outage_duration_min',  'sum'),
        mean_duration_min   = ('outage_duration_min',  'mean'),
        median_duration_min = ('outage_duration_min',  'median'),
        total_customers_aff = ('customers_affected',   'sum'),
        mean_demand_loss_mw = ('demand_loss_mw',       'mean'),
        mean_price_total    = ('price_total',          'mean'),
        mean_price_res      = ('price_residential',    'mean'),
        mean_price_com      = ('price_commercial',     'mean'),
        mean_price_ind      = ('price_industrial',     'mean'),
        mean_real_gsp       = ('real_gsp_usd',         'mean'),
    )
    .reset_index()
)

# Join state full name and climate region back for labeling
state_meta = (outage[['state_abbr','state','climate_region','nerc_region']]
              .drop_duplicates('state_abbr')
              .dropna(subset=['state_abbr']))
outage_agg = outage_agg.merge(state_meta, on='state_abbr', how='left')

print("AGGREGATED OUTAGE SHAPE")
print(outage_agg.shape)
outage_agg.head()
```
