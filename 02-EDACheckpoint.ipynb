## Data

### Data Overview

Our research question has two distinct components: (1) whether air pollutant levels predict power outage severity, and (2) whether these predictors vary by region. Answering it requires one dataset that measures ambient air pollution across U.S. states over time, and a separate dataset that records the severity, cause, and geographic location of power outage events over the same period. The CACES data has no outage information, and the Purdue data has no air quality measurements, so both are strictly necessary for this analysis. The analysis depends on merging them into a single state × year dataset where both pollution levels and outage metrics are jointly observed.

**CACES U.S. Air Quality Data** (https://www.caces.us/data) provides the air pollution side of the analysis. It supplies annual, county-level, population-weighted ambient concentration estimates forten pollutants — CO, NO₂, O₃, BC, PM10, PM2.5, SO₂, COA, HOA, and PNC — covering 1979–2020. Key columns are `fips` (county identifier), `pollutant`, `year`, `pred_wght` (the population-weighted concentration estimate), `state_abbr`, `lat`, and `lon`. Estimates are derived from EPA monitoring networks combined with land-use regression models, providing coverage even for counties without physical monitoring stations. Population-weighting means the estimate reflects the exposure experienced by an average resident of that county rather than a simple geographic average, making it more meaningful for health and policy analysis than a raw spatial mean.

Because the outage dataset records location only at the state level, we aggregate CACES county estimates to annual state means using `state_abbr` and `year` as keys. This loses within-state spatial variation but produces a common geographic identifier for merging. Four of the ten pollutants (BC, COA, HOA, PNC) have sparse coverage that does not overlap with the outage dataset's 2000–2016 window after merging, and are excluded from modeling — leaving CO, NO₂, O₃, PM10, PM2.5, and SO₂ as the six predictors used in analysis. These six are also the primary components of the EPA's AQI described in the background, making them the most policy-relevant measures available.

**Purdue Major Power Outage Risks in the U.S.** (https://engineering.purdue.edu/LASCI/research-data/outages) provides the outage severity side of the analysis. It records 1,534 major power outage events in the contiguous U.S. from January 2000 to July 2016, with one row per event. The variables we use as outage severity outcomes are `OUTAGE.DURATION` (minutes) and `CUSTOMERS.AFFECTED` — directly measuring how long and how widely each outage impacted the grid. `DEMAND.LOSS.MW` provides a third severity measure reflecting peak grid stress. `CAUSE.CATEGORY` is critical for our analysis because severe weather is the only cause with a plausible atmospheric connection to air quality — we use it to subset analyses and test whether the pollutant-outage association is stronger in weather-caused events specifically. `CLIMATE.REGION` (nine U.S. climate zones) is the primary variable for the regional heterogeneity component of our question, allowing us to test whether pollutant-outage associations differ across climate zones. Sector-level electricity prices (`RES.PRICE`, `COM.PRICE`, `IND.PRICE`, `TOTAL.PRICE`) serve as economic control variables rather than outcomes, since they reflect state energy market conditions rather than outage-specific severity.

Because outage events are recorded individually but CACES is annual and state-level, we aggregate outage events to state × year means and counts before merging. The two datasets are joined on `state_abbr` and `year` using an inner join, retaining only state-years present in restricting the analysis to 2000–2016 across the contiguous 48 states. The merged dataset is the sole input to all EDA and modeling: CACES contributes the six pollutant concentration columns, and the Purdue data contributes the outage severity metrics, cause breakdown, and climate region classification.

```python
# Missing data 
print("missing values")
miss = caces_raw.isnull().sum()
miss_pct = (miss / len(caces_raw) * 100).round(2)
missing_df = pd.DataFrame({'missing_count': miss, 'missing_pct': miss_pct})
print(missing_df[missing_df['missing_count'] > 0])

# Check if missingness is systematic by pollutant
print("\nmissing pred_wght by pollutant")
print(
    caces_raw[caces_raw['pred_wght'].isna()]
    .groupby('pollutant')
    .size()
    .rename('missing_count')
)
```

```python
# Outlier check on pred_wght 
print("pred_wght summary stats by pollutant")
print(
    caces_raw.groupby('pollutant')['pred_wght']
    .describe()
    .round(4)
    .to_string()
)

# Flag extreme outliers (> 5 IQR from median) per pollutant
def flag_outliers(group):
    q1, q3 = group['pred_wght'].quantile([0.25, 0.75])
    iqr = q3 - q1
    return group[
        (group['pred_wght'] < q1 - 5 * iqr) |
        (group['pred_wght'] > q3 + 5 * iqr)
    ]

outliers = caces_raw.groupby('pollutant', group_keys=False).apply(flag_outliers)
print(f"\nExtreme outliers (>5 IQR): {len(outliers)} rows")
print(outliers[['fips','state_abbr','year','pollutant','pred_wght']]
      .sort_values('pred_wght', ascending=False).head(10))
```

```python
# Clean & aggregate to state × year 
caces = caces_raw.copy()
caces.columns = caces.columns.str.strip().str.lower()

# Restrict to the 6 well-covered pollutants
POLLUTANTS = ["co", "no2", "o3", "pm10", "pm25", "so2"]
caces_filtered = caces[caces['pollutant'].str.lower().isin(POLLUTANTS)].copy()
caces_filtered['pollutant'] = caces_filtered['pollutant'].str.lower()

# Aggregate to state × year: population-weighted mean across counties
state_yr_long = (
    caces_filtered
    .groupby(['year', 'state_abbr', 'pollutant'])['pred_wght']
    .mean()
    .reset_index()
)

state_yr = (
    state_yr_long
    .pivot_table(index=['year','state_abbr'], columns='pollutant', values='pred_wght')
    .reset_index()
)
state_yr.columns.name = None
state_yr = state_yr.rename(columns={p: f'poll_{p}' for p in POLLUTANTS})

# Normalise keys
state_yr['state_abbr'] = state_yr['state_abbr'].str.strip().str.upper()
state_yr['year'] = state_yr['year'].astype(int)

print("processed caces shape")
print(state_yr.shape)
print(f"\nYears: {state_yr['year'].min()} – {state_yr['year'].max()}")
print(f"States: {state_yr['state_abbr'].nunique()}")
print("\nheaderS")
state_yr.head()
```

### Section 3 — Pollutant–Outage Associations

We examine the core research question: do states and years with higher 
pollutant concentrations experience more severe outages? We use two 
complementary approaches:

- **A correlation heatmap** showing raw Pearson r between each 
  pollutant and each outage metric, alongside partial correlations 
  that control for year to isolate the genuine association from the 
  shared time trend
- **A scatter plot** of the two strongest individual predictors vs. 
  log outage duration, color-coded by year to make the time trend 
  visible

```python
# Figure 4: Correlation heatmap — raw and partial
# Compute both raw correlations and year-partialed correlations

outcome_cols = ['mean_duration_min', 'median_duration_min',
                'n_outages', 'total_customers_aff']
outcome_labels = {
    'mean_duration_min':   'Mean Duration',
    'median_duration_min': 'Median Duration',
    'n_outages':           '# Outages',
    'total_customers_aff': 'Customers Affected',
}

raw_data, partial_data = [], []
for poll in poll_cols:
    raw_row, partial_row = {}, {}
    for out in outcome_cols:
        sub = merged[[poll, out, 'year']].dropna()
        raw_r = sub[[poll, out]].corr().iloc[0,1]
        pr, _  = partial_r_year(sub, poll, out)
        raw_row[outcome_labels[out]]     = round(raw_r, 2)
        partial_row[outcome_labels[out]] = round(pr, 2)
    raw_data.append(raw_row)
    partial_data.append(partial_row)

raw_df     = pd.DataFrame(raw_data,
                           index=[p.replace('poll_','').upper() for p in poll_cols])
partial_df = pd.DataFrame(partial_data,
                           index=raw_df.index)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, df, title in zip(
        axes,
        [raw_df, partial_df],
        ['Raw Pearson r\n(includes shared time trend)',
         'Partial r (controlling for year)\n(removes time-trend confound)']):
    sns.heatmap(df, annot=True, fmt='.2f', center=0, vmin=-0.5, vmax=0.5,
                cmap='coolwarm', linewidths=0.4, ax=ax, annot_kws={'size':11})
    ax.set_title(title)
    ax.set_xlabel('Outage Metric')
    ax.set_ylabel('Pollutant')

fig.suptitle('Pollutant–Outage Correlations: Raw vs. Controlling for Year\n'
             '(positive = higher pollution → more severe outages)', y=1.02)
plt.tight_layout()
plt.show()
```

The raw correlations (left) are consistently positive between pollutants and duration metrics (r = 0.10–0.25), but shrink after partialing out year (right). This confirms that part of the raw association reflects the shared downward time trend — pollution fell while outage patterns fluctuated independently — rather than a genuine relationship. However, most partial correlations remain positive and the pattern is consistent across all six pollutants, suggesting a real if modest association remains after controlling for time.

The negative correlations with outage *count* likely reflect the time trend most strongly: outage counts peaked around 2011 while pollution had already been declining for a decade.

### Section 4 — Regional Heterogeneity

The second part of our research question asks whether the pollutant–
outage association varies by climate region. We use two approaches:

- **Within-region correlations**: computing the pollutant–duration 
  Pearson r separately for each of the nine U.S. climate regions to 
  see whether the association is consistent or region-specific
- **K-Nearest Neighbors classification by region**: treating 
  above/below-median outage duration as a binary label and testing 
  whether pollutant levels predict it, separately per region — regions 
  where accuracy exceeds chance are ones where pollution is 
  informative about severity

```python
# Figure 5: Within-region correlation heatmap
regions = merged['climate_region'].dropna().unique()
results = []

for poll in poll_cols:
    for region in regions:
        sub = merged[merged['climate_region'] == region][
            [poll, 'log_mean_duration', 'year']].dropna()
        if len(sub) >= 5:
            pr, pp = partial_r_year(sub, poll, 'log_mean_duration')
            results.append({
                'pollutant': poll.replace('poll_','').upper(),
                'region': region,
                'partial_r': round(pr, 2),
                'p': round(pp, 3),
                'n': len(sub)
            })

corr_regional = pd.DataFrame(results)
corr_pivot    = corr_regional.pivot(index='region', columns='pollutant',
                                     values='partial_r')

fig, ax = plt.subplots(figsize=(12, 6))
sns.heatmap(corr_pivot, annot=True, fmt='.2f', center=0,
            vmin=-0.6, vmax=0.6,
            cmap='coolwarm', linewidths=0.4, ax=ax, annot_kws={'size':10})
ax.set_title('Partial r (controlling for year): Pollutants vs. Log Outage Duration\n'
             'by Climate Region  —  positive = higher pollution → longer outages',
             pad=12)
ax.set_xlabel('Pollutant')
ax.set_ylabel('Climate Region')
plt.tight_layout()
plt.show()

# Summarise which regions show the strongest / weakest signal
mean_r_by_region = corr_pivot.mean(axis=1).sort_values(ascending=False)
print("Mean partial r across pollutants by region:")
print(mean_r_by_region.round(3).to_string())
```

The within-region correlations vary substantially across climate zones, directly supporting the regional heterogeneity component of our hypothesis. Regions with consistent positive correlations across pollutants are ones where the atmospheric confounding mechanism is most plausible — high-pollution air masses and grid-stressing weather conditions co-occur there more reliably. Regions with near-zero or negative correlations suggest that local outage causes are driven by factors largely independent of ambient air quality.

We now test whether this regional variation translates into classification performance — can pollutant levels actually predict which state-years will have above-median outage duration, and does that predictive power differ by region?

```python
# Figure 6: KNN classification accuracy by region
# Binary label: above/below median log outage duration
median_dur = merged['log_mean_duration'].median()
merged['high_severity'] = (merged['log_mean_duration'] > median_dur).astype(int)

feature_cols = poll_cols + ['year']
cv           = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Overall accuracy across k to choose best k
clf_all  = merged[feature_cols + ['high_severity']].dropna()
X_all    = StandardScaler().fit_transform(clf_all[feature_cols])
y_all    = clf_all['high_severity'].values

k_scores = [cross_val_score(KNeighborsClassifier(n_neighbors=k),
                             X_all, y_all, cv=cv,
                             scoring='accuracy').mean()
            for k in range(1, 16)]
best_k = np.argmax(k_scores) + 1
overall_acc = max(k_scores)
print(f"Best k={best_k}, overall CV accuracy={overall_acc:.3f}")

# Per-region accuracy at best_k
region_accs = {}
clf_data = merged[feature_cols + ['high_severity','climate_region']].dropna()

for region in clf_data['climate_region'].unique():
    sub = clf_data[clf_data['climate_region'] == region]
    if len(sub) < 15:
        continue
    Xr  = StandardScaler().fit_transform(sub[feature_cols])
    yr  = sub['high_severity'].values
    k_r = min(best_k, len(sub) // 5)
    acc = cross_val_score(KNeighborsClassifier(n_neighbors=k_r),
                          Xr, yr, cv=StratifiedKFold(
                              n_splits=5, shuffle=True, random_state=42),
                          scoring='accuracy').mean()
    region_accs[region] = {'accuracy': acc, 'n': len(sub)}

region_acc_df = (pd.DataFrame(region_accs).T
                 .sort_values('accuracy', ascending=False))

fig, ax = plt.subplots(figsize=(11, 4))
bar_colors = ['tomato' if v > overall_acc else 'steelblue'
              for v in region_acc_df['accuracy']]
ax.bar(region_acc_df.index, region_acc_df['accuracy'],
       color=bar_colors, edgecolor='white')
ax.axhline(0.5, color='gray', ls=':', lw=1.5, label='Chance (0.50)')
ax.axhline(overall_acc, color='black', ls='--', lw=1.5,
           label=f'Overall CV accuracy ({overall_acc:.3f})')
ax.set_ylabel('5-Fold CV Accuracy')
ax.set_title('KNN Classification Accuracy by Climate Region\n'
             '(predicting above/below median outage duration from pollutant levels + year;\n'
             'red bars = above overall average, blue = below)', pad=10)
ax.tick_params(axis='x', rotation=40)
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()

print("\nPer-region results:")
print(region_acc_df.round(3).to_string())
```

### Section 5 — Weather-Caused vs. Non-Weather Outages: Does the Atmospheric Mechanism Hold?

Figure 3 established that severe weather dominates outage causes and produces the longest durations. The background section argues that the most plausible explanation for any pollutant–outage association is a shared atmospheric driver — stagnant high-pressure systems that simultaneously trap pollutants and stress the grid through heat and
wind. If this mechanism is correct, the pollutant–duration association should be meaningfully stronger in state-years where weather-caused outages dominate than in state-years where other causes (intentional attacks, equipment failure) dominate.

We test this by splitting the merged dataset into weather-dominated state-years (more than 50% of outages caused by severe weather) and non-weather-dominated state-years, then computing partial correlations — controlling for year — separately in each subset. Stronger and more significant correlations in the weather subset would support the atmospheric confounding hypothesis; similar correlations in both subsets would suggest the association operates through a different mechanism entirely.

```python
# Figure 7: Partial r by cause subset
# Add weather fraction back to merged
weather_frac = (outage[outage['weather_caused']]
                .groupby(['year','state_abbr'])
                .size()
                .reset_index(name='n_weather'))
total_out = (outage.groupby(['year','state_abbr'])
             .size()
             .reset_index(name='n_total'))
weather_info = weather_frac.merge(total_out, on=['year','state_abbr'])
weather_info['pct_weather'] = weather_info['n_weather'] / weather_info['n_total']

merged_w = merged.merge(weather_info[['year','state_abbr','pct_weather']],
                        on=['year','state_abbr'], how='left')
merged_w['pct_weather'] = merged_w['pct_weather'].fillna(0)

weather_sub    = merged_w[merged_w['pct_weather'] > 0.5].copy()
non_weather_sub = merged_w[merged_w['pct_weather'] <= 0.5].copy()

subsets = {
    'Weather-dominated\nstate-years': weather_sub,
    'Non-weather-dominated\nstate-years': non_weather_sub,
}

fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

for ax, (label, sub_df) in zip(axes, subsets.items()):
    r_vals, p_vals, poll_names = [], [], []
    for poll in poll_cols:
        r, p = partial_r_year(sub_df, poll, 'log_mean_duration')
        r_vals.append(r)
        p_vals.append(p)
        poll_names.append(poll.replace('poll_','').upper())

    bar_colors = ['tomato' if r > 0 else 'steelblue' for r in r_vals]
    bars = ax.bar(poll_names, r_vals, color=bar_colors, edgecolor='white')
    ax.axhline(0, color='gray', lw=0.8)
    ax.set_ylim(-0.5, 0.5)
    ax.set_xlabel('Pollutant')
    ax.set_ylabel('Partial r (controlling for year)')
    ax.set_title(f'{label}\n(n = {len(sub_df)} state-years)')

    # Annotate significance
    for bar, p in zip(bars, p_vals):
        sig = '***' if p<0.001 else '**' if p<0.01 else '*' if p<0.05 else ''
        if sig:
            ax.text(bar.get_x() + bar.get_width()/2,
                    bar.get_height() + 0.02, sig,
                    ha='center', fontsize=11, color='black')

fig.suptitle('Partial Correlations: Pollutants vs. Log Outage Duration\n'
             'Comparing Weather-Dominated vs. Non-Weather State-Years\n'
             '(* p<0.05, ** p<0.01, *** p<0.001)', y=1.02)
plt.tight_layout()
plt.show()

print("Mean partial r across pollutants:")
for label, sub_df in subsets.items():
    rs = [partial_r_year(sub_df, p, 'log_mean_duration')[0] for p in poll_cols]
    print(f"  {label.strip()}: {np.mean(rs):.3f}")
```

### Section 6 — Geographic Profile of Outage Severity and PM2.5

The regional correlation heatmap in Figure 5 shows how the statistical association between pollutants and duration varies by climate zone, but it does not show what the actual severity levels look like state by state. A geographic profile is useful for two reasons: it makes the regional clustering of outage severity concrete rather than numerical, and it lets us visually assess whether PM2.5 — our strongest individual predictor — tracks the duration pattern within regions or diverges from it.

We plot mean log outage duration as bars for each state, colored by climate region and sorted first by region (ordered by mean duration, highest to lowest) then by duration within each region. Mean PM2.5 is overlaid as a line on a secondary axis. If PM2.5 reliably tracks duration within regions, we would expect the black line to roughly follow the bar heights within each color group. Systematic divergence between the two would indicate that factors other than pollution — grid age, population density, local weather severity — are the primary drivers of the state-level outage profile.

```python
# Figure 8: State severity profile sorted by climate region
state_profile = (
    merged.groupby(['state','climate_region'])
    .agg(
        mean_log_dur = ('log_mean_duration', 'mean'),
        mean_pm25    = ('poll_pm25',         'mean'),
        n_years      = ('year',              'count'),
    )
    .reset_index()
    .dropna(subset=['mean_pm25','mean_log_dur','climate_region'])
)

# Sort by region then by duration within region
region_order = (state_profile.groupby('climate_region')['mean_log_dur']
                .mean().sort_values(ascending=False).index)
state_profile['region_rank'] = state_profile['climate_region'].map(
    {r: i for i, r in enumerate(region_order)})
state_profile = state_profile.sort_values(
    ['region_rank','mean_log_dur'], ascending=[True, False])

region_palette = dict(zip(region_order,
                          sns.color_palette('tab10', len(region_order))))
bar_colors = state_profile['climate_region'].map(region_palette)

fig, ax1 = plt.subplots(figsize=(16, 5))
ax2 = ax1.twinx()

x = np.arange(len(state_profile))
ax1.bar(x, state_profile['mean_log_dur'],
        color=bar_colors, edgecolor='white', lw=0.3, alpha=0.85,
        label='Log outage duration (left axis)')
ax2.plot(x, state_profile['mean_pm25'], color='black',
         marker='o', markersize=3, lw=1.2, alpha=0.6,
         label='Mean PM2.5 (right axis)')

ax1.set_xticks(x)
ax1.set_xticklabels(state_profile['state'], rotation=75,
                    ha='right', fontsize=7.5)
ax1.set_ylabel('Mean log(Outage Duration + 1)')
ax2.set_ylabel('Mean PM2.5 (µg/m³)')
ax1.set_title('Mean Log Outage Duration and PM2.5 by State\n'
              '(bars colored by climate region, sorted by region then duration;\n'
              'black line = PM2.5 — does it track outage severity within regions?)',
              pad=10)

# Legend for regions
legend_elements = [Patch(facecolor=region_palette[r], label=r)
                   for r in region_order]
ax1.legend(handles=legend_elements, loc='upper right',
           fontsize=7, ncol=2, title='Climate Region')

plt.tight_layout()
plt.show()
```

Shows mean log outage duration per state, grouped by climate region, alongside mean PM2.5 as a secondary axis. Sorting by region rather than magnitude makes regional clustering visible.

### EDA Summary

Six figures establish the following findings that directly motivate 
the predictive modeling to follow:

**1. All pollutants declined significantly 2000–2016** (Figure 1). 
This shared time trend is a confounder that must be controlled for — 
we include `year` as a covariate in all models.

**2. Outage duration is severely right-skewed** (Figure 2). 
Log-transformation brings skewness from >5 to near-zero and is used 
as the outcome variable throughout.

**3. Severe weather dominates outage causes** (Figure 3). 
It accounts for nearly half of all events and produces the longest 
durations. It is the only cause with a plausible atmospheric 
connection to air quality, motivating weather-cause subsetting in 
modeling.

**4. Pollutants show weak positive associations with outage duration, 
partially explained by the time trend** (Figure 4). Raw correlations 
(r = 0.10–0.25) shrink but remain positive after partialing out year, 
indicating a genuine if modest signal beyond the time confound.

**5. The association varies substantially by climate region** 
(Figure 5). Some regions show consistent positive partial correlations 
across all six pollutants; others show near-zero or mixed-sign 
results, directly supporting the regional heterogeneity component of 
our hypothesis.

**6. KNN classification accuracy varies by region** (Figure 6), 
confirming that pollutant levels are meaningfully more predictive of 
outage severity in some climate zones than others — and that this 
difference is detectable with a nonparametric classifier, not just 
correlation coefficients.

**7. The pollutant–outage association is stronger in weather-dominated
state-years than non-weather-dominated ones** (Figure 7). Partial
correlations are consistently larger in magnitude in state-years where
severe weather accounts for the majority of outages, and statistical
significance is more frequently achieved in that subset. This supports
the atmospheric confounding hypothesis from the background section:
the same stagnant high-pressure systems that trap pollutants also
cause the heat and wind conditions that stress grid infrastructure.
The association is not simply an artifact of industrial geography —
it is specifically amplified in the weather context where a shared
atmospheric driver is most plausible.

**8. Outage severity and PM2.5 do not track each other cleanly
within regions** (Figure 8). When states are sorted by climate region
and duration is plotted alongside PM2.5, the PM2.5 line does not
consistently mirror the duration bars within any single region. This
tempers the correlation findings from Figures 4 and 5 — while
statistical associations exist at the state-year level, they are not
strong enough to produce a visually obvious geographic pattern. This
is consistent with our hypothesis that the relationship is real but
modest, and that pollution is a proxy for underlying atmospheric
conditions rather than a direct predictor of outage severity.
